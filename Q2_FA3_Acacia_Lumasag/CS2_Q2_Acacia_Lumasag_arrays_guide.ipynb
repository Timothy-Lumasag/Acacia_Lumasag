## 📦 Part 1: What is an Array / List?

Imagine you are managing an online game inventory, a class record, or a music playlist.
- **Without Arrays:** You would need separate variables like `item1 = "Notebook"`, `item2 = "Pen"`, `item3 = "Calculator"`, etc. This gets messy very quickly!
- **With Arrays / Lists:** You can group all related items into **one single variable**:
  ```python
  backpack = ["Notebook", "Pen", "Calculator", "Laptop", "Book"]
  ```

### 💡 Real-World Analogy: A Row of Boxes
Think of a list as a row of numbered boxes sitting on a shelf. Each box holds **one item**, and each box has a position number called an **index**.

Let's run the code cell below to create our first Python lists!

````python
# Creating our first Python Lists (Datasets from Study Guide)

# Dataset A: School Backpack Items (Strings)
backpack = ["Notebook", "Pen", "Calculator", "Laptop", "Book"]

# Dataset B: Class Quiz Scores (Numbers)
quiz_scores = [85, 90, 78, 92, 88]

# Dataset C: Favorite Foods (Strings)
favorite_foods = ["Pizza", "Burger", "Pasta", "Fried Chicken", "Ice Cream"]

print("Backpack Items:", backpack)
print("Quiz Scores:", quiz_scores)
print("Favorite Foods:", favorite_foods)
````

## 🔢 Part 2: Understanding 0-Based Indexing

## 🔄 Part 3: Array Traversal Using Loops

While direct indexing is great for single items, **traversal** means visiting every element in an array one by one.

We use `for` loops to make traversal fast, clean, and efficient!

### Two Ways to Loop Through a List:
1. **Looping by Item:** `for item in list_name:`
2. **Looping by Index:** `for i in range(len(list_name)):`

Let's test both methods below!

````python
# Array Traversal Examples

print("=== Method 1: Looping by Item ===")
for item in backpack:
    print(f"🎒 Item: {item}")

print("\n=== Method 2: Looping by Index (with Position Labels) ===")
for i in range(len(backpack)):
    print(f"Index {i} -> {backpack[i]}")

print("\n=== Processing Data During Traversal (Calculating Average Score) ===")
total_score = 0
for score in quiz_scores:
    total_score += score

average_score = total_score / len(quiz_scores)
print(f"Total Points: {total_score}")
print(f"Average Score: {average_score:.2f}")
````

````python
#Activity 2: Process Your Data (Input + Computation/Search with Arrays/Lists)

#do your Activity 2 here

# Activity 2: Array Traversal Using Loops

# Create your list here
favorite_artists = ['The 1975', 'CORTIS', 'keshi', 'The Smashing Pumpkins', 'beabadoobee', 'Oasis', 'Blur']


# === Method 1: Looping by Item ===
print("=== Method 1: Looping by Item ===")

# Loop through each artist in the list
for artist in favorite_artists:
    print(f"🎵 Artist: {artist}")


# === Method 2: Looping by Index (with Position Labels) ===
print("\n=== Method 2: Looping by Index (with Position Labels) ===")

# Loop through the list using its index
for i in range(len(favorite_artists)):
    print(f"Index {i} -> {favorite_artists[i]}")


# === Processing Data During Traversal ===
print("\n=== Processing Data During Traversal ===")

# Count the total number of favorite artists
total_artists = 0

# Loop through the list and count each artist
for artist in favorite_artists:
    total_artists += 1

# Display the total number of artists
print(f"Total Favorite Artists: {total_artists}")
````

## ✂️ Part 4: List Slicing & Searching

### 1. List Slicing (`list[start:end]`)
Slicing allows you to extract a sub-list (a portion of your list).
- Syntax: `list_name[start:end]`
- **Remember:** The slice starts at `start` index and goes up to, but **does NOT include**, `end` index!
- Example: `backpack[1:4]` gets items at index 1, 2, and 3.

### 2. Linear Searching
Searching means checking if a target value exists in your list and finding its index position.

````python
# Activity 3: Slicing and Searching

# Create your list here
favorite_artists = ['The 1975', 'CORTIS', 'keshi', 'The Smashing Pumpkins', 'beabadoobee', 'Oasis', 'Blur']

print("Original Favorite Artists List:", favorite_artists)


# Extracting a sliced portion from index 1 to 4
# Indexes 1, 2, and 3 will be included
sliced_artists = favorite_artists[1:4]

print("Sliced Sub-List (1:4):", sliced_artists)


# Linear Search Function
search_artist = "keshi"

# Check if the artist is in the sliced list
if search_artist in sliced_artists:
    found_index = sliced_artists.index(search_artist)
    print(f"🔍 Search Result: '{search_artist}' was found in the sliced list at sub-index {found_index}!")
else:
    print(f"🔍 Search Result: '{search_artist}' was not found in the sliced list.")
````

## 🧰 Part 5: Essential Python Built-in List Methods

Python provides powerful built-in methods and functions to modify and inspect lists without writing long loops!

| Method / Function | What It Does | Example |
| :--- | :--- | :--- |
| **`len(list)`** | Returns total number of items | `len(backpack)` |
| **`sum(list)`** | Calculates sum of numeric items | `sum(quiz_scores)` |
| **`max(list)` / `min(list)`** | Finds highest / lowest value | `max(quiz_scores)` |
| **`append(item)`** | Adds item to the **end** of list | `backpack.append("Binder")` |
| **`insert(index, item)`** | Inserts item at specific position | `backpack.insert(1, "Ruler")` |
| **`remove(item)`** | Removes first occurrence of item | `backpack.remove("Pen")` |
| **`pop(index)`** | Removes & returns item at index | `backpack.pop(0)` |
| **`sort()` / `sorted()`** | Sorts list in ascending order | `quiz_scores.sort()` |
| **`reverse()`** | Reverses order of list items | `backpack.reverse()` |

Try running the code below to see these methods in action!

````python
# Demonstrating Built-in Methods

scores = [85, 90, 78, 92, 88]
print("Initial Scores:", scores)

# Math Functions
print(f"Total Items: {len(scores)}")
print(f"Sum of Scores: {sum(scores)}")
print(f"Highest Score: {max(scores)}")
print(f"Lowest Score: {min(scores)}")

# Modifying List
scores.append(95)
print("After append(95):", scores)

scores.sort()
print("After sort():", scores)

scores.reverse()
print("After reverse():", scores)
````

## 🏆 Part 7: Mini-Project — Student Score Analyzer Mini-App

Create a complete modular text-based program that manages and analyzes student scores!

This program uses a **modular structure with functions** and a clear **interactive menu** matching Activity 3 (Data Analysis Mini-App).

Run the cell below to test the Student Score Analyzer!

````python
# Student Score Analyzer Mini-App

scores_data = [85, 90, 78, 92, 88]

def display_scores(scores):
    print("\n--- Current Student Scores ---")
    for i, s in enumerate(scores):
        print(f"Student #{i+1} (Index {i}): {s}")

def search_score(scores, target):
    if target in scores:
        idx = scores.index(target)
        print(f"✅ Score {target} found at Student #{idx+1} (Index {idx}).")
    else:
        print(f"❌ Score {target} not found in records.")

def compute_statistics(scores):
    if not scores:
        print("No scores available.")
        return
    avg = sum(scores) / len(scores)
    print(f"\n--- Score Statistics ---")
    print(f"Total Students: {len(scores)}")
    print(f"Average Score:  {avg:.2f}")
    print(f"Highest Score:  {max(scores)}")
    print(f"Lowest Score:   {min(scores)}")

def update_score(scores, index, new_score):
    if 0 <= index < len(scores):
        old_val = scores[index]
        scores[index] = new_score
        print(f"✅ Updated Index {index}: {old_val} -> {new_score}")
    else:
        print(f"❌ Invalid index! Must be between 0 and {len(scores)-1}.")

def run_mini_app():
    print("=== Welcome to Student Score Analyzer ===")

    # Demonstration Run
    display_scores(scores_data)
    compute_statistics(scores_data)

    print("\n--- Performing Search ---")
    search_score(scores_data, 92)

    print("\n--- Updating Score ---")
    update_score(scores_data, 2, 82) # Update 78 to 82

    display_scores(scores_data)
    compute_statistics(scores_data)

# Run the Mini-App
run_mini_app()
````

````python
#Activity 3 (Graded or Non-Graded Formative Assessment): Data Analysis Mini App

# Do your Activity 3 here

# Student Score Analyzer Mini-App

# Create your list of student scores
scores_data = [85, 90, 78, 92, 88]


# === Display Student Scores ===
def display_scores(scores):
    print("\n--- Current Student Scores ---")

    # Loop through the scores and display each student
    for i, score in enumerate(scores):
        print(f"Student #{i+1} (Index {i}): {score}")


# === Search for a Score ===
def search_score(scores, target):
    # Check if the target score exists
    if target in scores:
        idx = scores.index(target)
        print(f"✅ Score {target} found at Student #{idx+1} (Index {idx}).")
    else:
        print(f"❌ Score {target} not found in records.")


# === Compute Score Statistics ===
def compute_statistics(scores):
    # Check if there are no scores
    if not scores:
        print("No scores available.")
        return

    # Calculate the average score
    average = sum(scores) / len(scores)

    print("\n--- Score Statistics ---")
    print(f"Total Students: {len(scores)}")
    print(f"Average Score:  {average:.2f}")
    print(f"Highest Score:  {max(scores)}")
    print(f"Lowest Score:   {min(scores)}")


# === Update a Student Score ===
def update_score(scores, index, new_score):
    # Check if the index is valid
    if 0 <= index < len(scores):
        old_score = scores[index]
        scores[index] = new_score
        print(f"✅ Updated Index {index}: {old_score} -> {new_score}")
    else:
        print(f"❌ Invalid index! Must be between 0 and {len(scores)-1}.")


# === Add a New Student Score ===
def add_score(scores, new_score):
    # Add the new score to the list
    scores.append(new_score)
    print(f"✅ Score {new_score} added successfully.")


# === Main Interactive Menu ===
def run_mini_app():
    print("=== Welcome to Student Score Analyzer ===")

    # Display the initial student scores
    display_scores(scores_data)

    # Display the initial statistics
    compute_statistics(scores_data)

    # Interactive menu
    while True:
        print("\n--- Student Score Analyzer Menu ---")
        print("1. Display Student Scores")
        print("2. Search for a Score")
        print("3. View Score Statistics")
        print("4. Update a Score")
        print("5. Add a New Score")
        print("6. Exit")

        choice = input("Enter your choice: ")

        # === Option 1: Display Scores ===
        if choice == "1":
            display_scores(scores_data)

        # === Option 2: Search Score ===
        elif choice == "2":
            target = int(input("Enter score to search: "))
            search_score(scores_data, target)

        # === Option 3: Statistics ===
        elif choice == "3":
            compute_statistics(scores_data)

        # === Option 4: Update Score ===
        elif choice == "4":
            index = int(input("Enter student index to update: "))
            new_score = int(input("Enter new score: "))
            update_score(scores_data, index, new_score)

        # === Option 5: Add Score ===
        elif choice == "5":
            new_score = int(input("Enter new student score: "))
            add_score(scores_data, new_score)

        # === Option 6: Exit ===
        elif choice == "6":
            print("👋 Thank you for using Student Score Analyzer!")
            break

        # === Invalid Menu Choice ===
        else:
            print("❌ Invalid choice! Please select 1-6.")


# === Run the Mini-App ===
run_mini_app()
````

````python
# Practice ^^ (On my own maam)

# Artist Song Analyzer Mini-app

# Artist Song Analyzer Mini-App

# === Artist Data ===
# Each artist has a list of songs under them.
# You can edit the artist names and song titles here.

artists_data = {
    "The 1975": ["YAMTS", "About You", "fallingforyou", "im in love with you", "Robbers"],
    "CORTIS": ["JoyRide", "Wassup", "GO!", "FaSHioN", "Acai"],
    "keshi": ["UNDERSTAND", "Soft Spot", "2 soon", "LIMBO", "always"],
    "The Smashing Pumpkins": ["1979", "Luna", "Mayonaise", "MCATIS", "Tonight, Tonight"],
    "beabadoobee": ["Everything I Want", "Lovesong", "AIDWDOY", "Everest", "The Perfect Pair"],
    "Oasis": ["Champagne Supernova", "Wonderwall", "Don't Look Back in Anger", "Stop Crying Your Heart Out", "Stand by Me"],
    "Blur": ["Tender", "Song 2", "Beetlebum", "Girls & Boys", "Coffee & TV"]
}


# === Display All Artists ===
def display_artists(artists):
    print("\n--- Favorite Artists ---")

    # Loop through every artist
    for i, artist in enumerate(artists):
        print(f"{i}. {artist}")


# === Display Songs Under an Artist ===
def display_songs(artists, artist_index):
    # Convert dictionary keys into a list
    artist_list = list(artists.keys())

    # Check if the artist index is valid
    if 0 <= artist_index < len(artist_list):

        artist = artist_list[artist_index]
        songs = artists[artist]

        print(f"\n--- Songs by {artist} ---")

        # Display each song with its index
        for i, song in enumerate(songs):
            print(f"Song #{i+1} (Index {i}): {song}")

    else:
        print("❌ Invalid artist index.")


# === Search for an Artist ===
def search_artist(artists, target):
    # Check if the artist exists
    if target in artists:
        artist_list = list(artists.keys())
        index = artist_list.index(target)

        print(f"✅ Artist '{target}' found at Index {index}.")
        display_songs(artists, index)

    else:
        print(f"❌ Artist '{target}' not found.")


# === Update a Song ===
def update_song(artists, artist_index, song_index, new_song):
    # Convert dictionary keys into a list
    artist_list = list(artists.keys())

    # Check if artist index is valid
    if 0 <= artist_index < len(artist_list):

        artist = artist_list[artist_index]
        songs = artists[artist]

        # Check if song index is valid
        if 0 <= song_index < len(songs):

            old_song = songs[song_index]
            songs[song_index] = new_song

            print(f"✅ Updated '{artist}' - Index {song_index}:")
            print(f"   {old_song} -> {new_song}")

        else:
            print(
                f"❌ Invalid song index! "
                f"Must be between 0 and {len(songs)-1}."
            )

    else:
        print("❌ Invalid artist index.")


# === Add a New Song ===
def add_song(artists, artist_index, new_song):
    # Convert dictionary keys into a list
    artist_list = list(artists.keys())

    # Check if artist index is valid
    if 0 <= artist_index < len(artist_list):

        artist = artist_list[artist_index]

        # Add the song under the selected artist
        artists[artist].append(new_song)

        print(f"✅ '{new_song}' added to {artist}.")

    else:
        print("❌ Invalid artist index.")


# === Count Artists and Songs ===
def compute_statistics(artists):
    # Count the number of artists
    total_artists = len(artists)

    # Count the total number of songs
    total_songs = 0

    for songs in artists.values():
        total_songs += len(songs)

    print("\n--- Artist & Song Statistics ---")
    print(f"Total Artists: {total_artists}")
    print(f"Total Songs:   {total_songs}")


# === Main Mini-App ===
def run_mini_app():
    print("=== Welcome to Artist Song Analyzer ===")

    # Display the artists when the program starts
    display_artists(artists_data)

    # Display initial statistics
    compute_statistics(artists_data)

    # Interactive Menu
    while True:

        print("\n--- Artist Song Analyzer Menu ---")
        print("1. Display Artists")
        print("2. Select an Artist")
        print("3. Search for an Artist")
        print("4. Change a Song")
        print("5. Add a Song")
        print("6. View Statistics")
        print("7. Exit")

        choice = input("Enter your choice: ")

        # === Option 1: Display Artists ===
        if choice == "1":
            display_artists(artists_data)


        # === Option 2: Select an Artist ===
        elif choice == "2":

            display_artists(artists_data)

            artist_index = int(input("Enter the artist index: "))

            display_songs(artists_data, artist_index)


        # === Option 3: Search for an Artist ===
        elif choice == "3":

            artist = input("Enter artist name to search: ")

            search_artist(artists_data, artist)


        # === Option 4: Change a Song ===
        elif choice == "4":

            display_artists(artists_data)

            artist_index = int(input("\nEnter artist index: "))

            display_songs(artists_data, artist_index)

            song_index = int(input("\nEnter song index to change: "))

            new_song = input("Enter the new song name: ")

            update_song(
                artists_data,
                artist_index,
                song_index,
                new_song
            )


        # === Option 5: Add a Song ===
        elif choice == "5":

            display_artists(artists_data)

            artist_index = int(input("\nEnter artist index: "))

            new_song = input("Enter the new song name: ")

            add_song(
                artists_data,
                artist_index,
                new_song
            )


        # === Option 6: View Statistics ===
        elif choice == "6":

            compute_statistics(artists_data)


        # === Option 7: Exit ===
        elif choice == "7":

            print("\n👋 Thank you for using Artist Song Analyzer!")
            break


        # === Invalid Choice ===
        else:

            print("❌ Invalid choice! Please select 1-7.")


# === Run the Mini-App ===
run_mini_app()
````
